```python
# pip install pandas numpy scikit-learn

from pathlib import Path
import numpy as np
import pandas as pd

RAW_PATH = Path("sl_electricity_load.csv")
CLEAN_PATH = Path("data/processed/sl_electricity_load_clean.csv")


def get_sri_lanka_monsoon(month: int) -> str:
    """Map month (1-12) to Sri Lanka's four climate/monsoon seasons."""
    if month in (3, 4):
        return "First_Inter_Monsoon"
    if month in (5, 6, 7, 8, 9):
        return "Southwest_Monsoon"
    if month in (10, 11):
        return "Second_Inter_Monsoon"
    return "Northeast_Monsoon"


def clean_currency_or_numeric(series: pd.Series) -> pd.Series:
    """Clean strings like 'LKR 32.50' or '12,400 kW' into float."""
    if pd.api.types.is_numeric_dtype(series):
        return series.astype(float)
    cleaned = (
        series.astype(str)
        .str.replace(r"LKR|Rs\.?|kW|MW|,", "", regex=True)
        .str.strip()
    )
    return pd.to_numeric(cleaned, errors="coerce")


def load_and_clean_data(raw_path: Path = RAW_PATH, save_path: Path = CLEAN_PATH) -> pd.DataFrame:
    if not raw_path.exists():
        raise FileNotFoundError(
            f"Raw dataset not found at '{raw_path}'. "
            "Place your Sri Lankan electricity CSV there before running."
        )

    df = pd.read_csv(raw_path)

    # 1. Standardize common column names (adjust if your CSV uses different headers)
    rename_map = {
        "Timestamp": "timestamp",
        "Load Demand (kW)": "load_kw",
        "Temperature (°C)": "temperature_c",
        "Humidity (%)": "humidity_pct",
        "Rainfall (mm)": "rainfall_mm",
        "Solar Irradiance (W/m²)": "solar_irradiance",
        "Electricity Price (LKR/kWh)": "tariff_lkr_kwh",
        "Public Event": "public_event",
    }
    df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})

    # 2. Parse timestamp, deduplicate, and sort chronologically
    df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")
    df = df.dropna(subset=["timestamp"]).drop_duplicates(subset=["timestamp"])
    df = df.sort_values("timestamp").set_index("timestamp")

    # 3. Clean numeric / LKR text columns
    numeric_cols = [
        c for c in ["load_kw", "temperature_c", "humidity_pct", "rainfall_mm", "solar_irradiance", "tariff_lkr_kwh"]
        if c in df.columns
    ]
    for col in numeric_cols:
        df[col] = clean_currency_or_numeric(df[col])

    # 4. Handle missing values in numeric features via time interpolation + forward-fill
    for col in numeric_cols:
        df[col] = df[col].interpolate(method="time", limit_direction="forward").ffill()

    # 5. Outlier inspection & gentle capping on non-negative load (only if positive load exists)
    if "load_kw" in df.columns:
        df.loc[df["load_kw"] < 0, "load_kw"] = np.nan
        df["load_kw"] = df["load_kw"].interpolate(method="time").ffill()
        q01, q99 = df["load_kw"].quantile([0.001, 0.999])
        df["is_extreme_load"] = ((df["load_kw"] < q01) | (df["load_kw"] > q99)).astype(int)
        df["load_kw_clean"] = df["load_kw"].clip(lower=q01, upper=q99)

    # 6. Calendar & Sri Lankan monsoon features
    df["hour"] = df.index.hour
    df["day_of_week"] = df.index.dayofweek
    df["month"] = df.index.month
    df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)
    df["monsoon_season"] = df["month"].map(get_sri_lanka_monsoon)

    # 7. Past-only lag and rolling features for forecasting
    target_col = "load_kw_clean" if "load_kw_clean" in df.columns else "load_kw"
    for lag in [1, 4, 24, 96, 168]:
        df[f"{target_col}_lag_{lag}"] = df[target_col].shift(lag)

    df[f"{target_col}_roll_mean_24"] = (
        df[target_col].shift(1).rolling(window=24, min_periods=1).mean()
    )
    df[f"{target_col}_roll_std_24"] = (
        df[target_col].shift(1).rolling(window=24, min_periods=1).std().fillna(0.0)
    )

    # 8. Encode categorical columns cleanly (one-hot for monsoon season)
    monsoon_dummies = pd.get_dummies(df["monsoon_season"], prefix="monsoon", dtype=int)
    df = pd.concat([df, monsoon_dummies], axis=1)

    # Drop rows with NaNs introduced by lagging at the start of the series
    df_model = df.dropna().copy()

    save_path.parent.mkdir(parents=True, exist_ok=True)
    df_model.to_csv(save_path)
    print(f"Saved cleaned dataset with {len(df_model):,} rows and {df_model.shape[1]} columns to {save_path}")
    return df_model


if __name__ == "__main__":
    load_and_clean_data()
```

```python
# pip install pandas numpy matplotlib seaborn statsmodels

from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.seasonal import seasonal_decompose

CLEAN_PATH = Path("data/processed/sl_electricity_load_clean.csv")
FIG_DIR = Path("reports/figures")


def run_eda(clean_path: Path = CLEAN_PATH, fig_dir: Path = FIG_DIR) -> None:
    if not clean_path.exists():
        raise FileNotFoundError(
            f"Cleaned CSV not found at '{clean_path}'. Run Step 2 first."
        )

    fig_dir.mkdir(parents=True, exist_ok=True)
    sns.set_theme(style="whitegrid")

    df = pd.read_csv(clean_path, index_col=0, parse_dates=True).sort_index()
    target = "load_kw_clean" if "load_kw_clean" in df.columns else "load_kw"

    print("Rows:", len(df), "| Date range:", df.index.min(), "to", df.index.max())
    print(df[[c for c in [target, "temperature_c", "humidity_pct", "rainfall_mm"] if c in df.columns]].describe().round(2))

    # 1. Full time-series line plot
    fig, ax = plt.subplots(figsize=(13, 4))
    ax.plot(df.index, df[target], linewidth=0.8, color="#1f77b4")
    ax.set_title("Sri Lankan Electricity Load Over Time")
    ax.set_xlabel("Timestamp")
    ax.set_ylabel(target)
    fig.tight_layout()
    fig.savefig(fig_dir / "01_load_timeseries.png", dpi=150)
    plt.close(fig)

    # 2. Seasonal decomposition on daily average load (period=365 if >= 2 years, else 30)
    daily_load = df[target].resample("D").mean().dropna()
    period = 365 if len(daily_load) >= 730 else max(7, min(30, len(daily_load) // 3))
    if len(daily_load) > period * 2:
        decomp = seasonal_decompose(daily_load, model="additive", period=period)
        fig = decomp.plot()
        fig.set_size_inches(11, 7)
        fig.tight_layout()
        fig.savefig(fig_dir / "02_seasonal_decomposition.png", dpi=150)
        plt.close(fig)

    # 3. Hourly & day-of-week demand patterns
    if "hour" in df.columns and "day_of_week" in df.columns:
        fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
        df.groupby("hour")[target].mean().plot(kind="bar", ax=axes[0], color="#2ca02c")
        axes[0].set_title("Average Load by Hour of Day")
        axes[0].set_xlabel("Hour")
        axes[0].set_ylabel(target)

        day_labels = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
        dow_means = df.groupby("day_of_week")[target].mean()
        axes[1].bar([day_labels[i] for i in dow_means.index], dow_means.values, color="#ff7f0e")
        axes[1].set_title("Average Load by Day of Week")
        axes[1].set_xlabel("Day of Week")
        axes[1].set_ylabel(target)

        fig.tight_layout()
        fig.savefig(fig_dir / "03_hourly_and_dow_patterns.png", dpi=150)
        plt.close(fig)

    # 4. Monsoon season boxplot
    if "monsoon_season" in df.columns:
        fig, ax = plt.subplots(figsize=(10, 4.5))
        sns.boxplot(data=df, x="monsoon_season", y=target, ax=ax)
        ax.set_title("Load Distribution Across Sri Lankan Monsoon Seasons")
        ax.set_xlabel("Monsoon Season")
        ax.set_ylabel(target)
        plt.xticks(rotation=15)
        fig.tight_layout()
        fig.savefig(fig_dir / "04_monsoon_boxplot.png", dpi=150)
        plt.close(fig)

    # 5. Autocorrelation (ACF) & Lag-1 scatter
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    max_lags = min(168, len(df) // 4)
    if max_lags >= 5:
        plot_acf(df[target].iloc[: min(len(df), 5000)], lags=max_lags, ax=axes[0])
        axes[0].set_title("Autocorrelation (ACF)")

    lag1_col = f"{target}_lag_1"
    if lag1_col in df.columns:
        sample_df = df[[lag1_col, target]].sample(min(len(df), 3000), random_state=42)
        axes[1].scatter(sample_df[lag1_col], sample_df[target], alpha=0.25, s=10)
        axes[1].set_title("Lag-1 vs. Current Load")
        axes[1].set_xlabel(lag1_col)
        axes[1].set_ylabel(target)

    fig.tight_layout()
    fig.savefig(fig_dir / "05_acf_and_lag_scatter.png", dpi=150)
    plt.close(fig)

    # 6. Correlation heatmap of core numeric features
    corr_cols = [
        c for c in [target, "temperature_c", "humidity_pct", "rainfall_mm", "solar_irradiance", "tariff_lkr_kwh", "hour", "is_weekend"]
        if c in df.columns
    ]
    if len(corr_cols) >= 2:
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.heatmap(df[corr_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax)
        ax.set_title("Feature Correlation Heatmap")
        fig.tight_layout()
        fig.savefig(fig_dir / "06_correlation_heatmap.png", dpi=150)
        plt.close(fig)

    print(f"EDA plots saved to {fig_dir}")


if __name__ == "__main__":
    run_eda()
```

```python
# pip install pandas numpy scikit-learn xgboost lightgbm catboost statsmodels joblib tqdm

import time
from pathlib import Path
import joblib
import numpy as np
import pandas as pd
from catboost import CatBoostRegressor
from lightgbm import LGBMRegressor
from sklearn.base import clone
from sklearn.ensemble import BaggingRegressor, RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from statsmodels.tsa.arima.model import ARIMA
from tqdm.auto import tqdm
from xgboost import XGBRegressor

SEED = 42
CLEAN_PATH = Path("data/processed/sl_electricity_load_clean.csv")
ARTIFACT_DIR = Path("artifacts/models")
MAX_TRAIN_ROWS = 20000


def compute_metrics(y_true: np.ndarray, y_pred: np.ndarray, train_time_sec: float) -> dict:
    rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
    mae = float(mean_absolute_error(y_true, y_pred))
    r2 = float(r2_score(y_true, y_pred))
    non_zero = np.abs(y_true) > 1e-6
    mape = (
        float(np.mean(np.abs((y_true[non_zero] - y_pred[non_zero]) / y_true[non_zero])) * 100)
        if np.any(non_zero)
        else np.nan
    )
    return {
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "MAPE_%": round(mape, 3) if not np.isnan(mape) else np.nan,
        "R2": round(r2, 4),
        "Train_Time_s": round(train_time_sec, 2),
    }


def prepare_time_series_splits(
    df: pd.DataFrame,
    test_ratio: float = 0.2,
    max_train_rows: int = MAX_TRAIN_ROWS,
):
    target_col = "load_kw_clean" if "load_kw_clean" in df.columns else "load_kw"
    drop_cols = {
        c
        for c in ["load_kw", "load_kw_clean", "is_extreme_load", "monsoon_season"]
        if c in df.columns
    }
    feature_cols = [
        c
        for c in df.columns
        if c not in drop_cols and pd.api.types.is_numeric_dtype(df[c])
    ]

    split_idx = int(len(df) * (1.0 - test_ratio))
    train_df = df.iloc[:split_idx].copy()
    test_df = df.iloc[split_idx:].copy()

    if max_train_rows and len(train_df) > max_train_rows:
        train_df = train_df.iloc[-max_train_rows:].copy()

    X_train = train_df[feature_cols].copy()
    y_train = train_df[target_col].to_numpy(dtype=float)
    X_test = test_df[feature_cols].copy()
    y_test = test_df[target_col].to_numpy(dtype=float)

    return X_train, y_train, X_test, y_test, train_df, test_df, target_col, feature_cols


def fit_timeseries_stacker(base_models: dict, X_train: pd.DataFrame, y_train: np.ndarray, X_test: pd.DataFrame, tscv: TimeSeriesSplit):
    """Build a time-series safe stacker using expanding-window validation predictions + Ridge meta-model."""
    meta_X_parts = []
    meta_y_parts = []

    for tr_idx, val_idx in tscv.split(X_train):
        X_tr_fold = X_train.iloc[tr_idx]
        y_tr_fold = y_train[tr_idx]
        X_val_fold = X_train.iloc[val_idx]
        y_val_fold = y_train[val_idx]

        fold_preds = []
        for _, est in base_models.items():
            m = clone(est)
            m.fit(X_tr_fold, y_tr_fold)
            fold_preds.append(m.predict(X_val_fold))

        meta_X_parts.append(np.column_stack(fold_preds))
        meta_y_parts.append(y_val_fold)

    meta_X_train = np.vstack(meta_X_parts)
    meta_y_train = np.concatenate(meta_y_parts)

    meta_model = Ridge(alpha=10.0)
    meta_model.fit(meta_X_train, meta_y_train)

    fitted_base = {}
    test_base_preds = []
    for name, est in base_models.items():
        m = clone(est)
        m.fit(X_train, y_train)
        fitted_base[name] = m
        test_base_preds.append(m.predict(X_test))

    test_pred = meta_model.predict(np.column_stack(test_base_preds))
    return {"base_models": fitted_base, "meta_model": meta_model}, test_pred


def run_models(clean_path: Path = CLEAN_PATH, artifact_dir: Path = ARTIFACT_DIR):
    if not clean_path.exists():
        raise FileNotFoundError(f"Cleaned dataset not found at '{clean_path}'. Run Step 2 first.")

    artifact_dir.mkdir(parents=True, exist_ok=True)
    df = pd.read_csv(clean_path, index_col=0, parse_dates=True).sort_index()

    X_train, y_train, X_test, y_test, train_df, test_df, target_col, feature_cols = prepare_time_series_splits(df)
    print(
        f"Loaded {len(df):,} total rows | Using Train: {len(X_train):,} | Test: {len(X_test):,} | Features: {len(feature_cols)}"
    )

    tscv = TimeSeriesSplit(n_splits=3)
    predictions = {}
    metrics_rows = {}
    trained_models = {}

    steps = [
        "1/8 Baseline Linear Regression",
        "2/8 Bagging: Random Forest (CV)",
        "3/8 Bagging: BaggingRegressor",
        "4/8 Boosting: XGBoost (CV)",
        "5/8 Boosting: LightGBM (CV)",
        "6/8 Boosting: CatBoost",
        "7/8 Stacking Ensemble",
        "8/8 Hybrid ARIMA + RF Residual",
    ]

    pbar = tqdm(steps, desc="Training models", unit="model")

    for step_name in pbar:
        pbar.set_postfix_str(step_name)

        # 1. Baseline: Linear Regression
        if step_name.startswith("1/8"):
            t0 = time.perf_counter()
            baseline = make_pipeline(StandardScaler(), LinearRegression())
            baseline.fit(X_train, y_train)
            t1 = time.perf_counter()
            pred_base = baseline.predict(X_test)
            predictions["Baseline_LinearRegression"] = pred_base
            metrics_rows["Baseline_LinearRegression"] = compute_metrics(y_test, pred_base, t1 - t0)
            trained_models["baseline_lr"] = baseline
            tqdm.write(f"[1/8] Baseline done in {t1 - t0:.2f}s")

        # 2. Bagging A: Random Forest Regressor
        elif step_name.startswith("2/8"):
            t0 = time.perf_counter()
            rf_search = RandomizedSearchCV(
                estimator=RandomForestRegressor(random_state=SEED, n_jobs=-1),
                param_distributions={
                    "n_estimators": [60, 100],
                    "max_depth": [10, 16],
                    "min_samples_split": [2, 5, 10],
                    "max_features": ["sqrt", 0.7],
                },
                n_iter=3,
                cv=tscv,
                scoring="neg_root_mean_squared_error",
                random_state=SEED,
                n_jobs=1,
            )
            rf_search.fit(X_train, y_train)
            t1 = time.perf_counter()
            rf_model = rf_search.best_estimator_
            pred_rf = rf_model.predict(X_test)
            predictions["Bagging_RandomForest"] = pred_rf
            metrics_rows["Bagging_RandomForest"] = compute_metrics(y_test, pred_rf, t1 - t0)
            trained_models["random_forest"] = rf_model
            tqdm.write(f"[2/8] Random Forest done in {t1 - t0:.2f}s | Best: {rf_search.best_params_}")

        # 3. Bagging B: BaggingRegressor
        elif step_name.startswith("3/8"):
            t0 = time.perf_counter()
            bag_model = BaggingRegressor(
                n_estimators=50,
                max_samples=0.7,
                max_features=0.8,
                random_state=SEED,
                n_jobs=-1,
            )
            bag_model.fit(X_train, y_train)
            t1 = time.perf_counter()
            pred_bag = bag_model.predict(X_test)
            predictions["Bagging_Regressor"] = pred_bag
            metrics_rows["Bagging_Regressor"] = compute_metrics(y_test, pred_bag, t1 - t0)
            trained_models["bagging_regressor"] = bag_model
            tqdm.write(f"[3/8] BaggingRegressor done in {t1 - t0:.2f}s")

        # 4. Boosting A: XGBoost
        elif step_name.startswith("4/8"):
            t0 = time.perf_counter()
            xgb_search = RandomizedSearchCV(
                estimator=XGBRegressor(
                    objective="reg:squarederror",
                    random_state=SEED,
                    n_jobs=2,
                    tree_method="hist",
                ),
                param_distributions={
                    "n_estimators": [100, 180],
                    "learning_rate": [0.03, 0.05, 0.1],
                    "max_depth": [4, 6],
                    "subsample": [0.8, 1.0],
                    "colsample_bytree": [0.8, 1.0],
                },
                n_iter=3,
                cv=tscv,
                scoring="neg_root_mean_squared_error",
                random_state=SEED,
                n_jobs=1,
            )
            xgb_search.fit(X_train, y_train)
            t1 = time.perf_counter()
            xgb_model = xgb_search.best_estimator_
            pred_xgb = xgb_model.predict(X_test)
            predictions["Boosting_XGBoost"] = pred_xgb
            metrics_rows["Boosting_XGBoost"] = compute_metrics(y_test, pred_xgb, t1 - t0)
            trained_models["xgboost"] = xgb_model
            tqdm.write(f"[4/8] XGBoost done in {t1 - t0:.2f}s | Best: {xgb_search.best_params_}")

        # 5. Boosting B: LightGBM
        elif step_name.startswith("5/8"):
            t0 = time.perf_counter()
            lgbm_search = RandomizedSearchCV(
                estimator=LGBMRegressor(
                    random_state=SEED,
                    n_jobs=2,
                    verbose=-1,
                ),
                param_distributions={
                    "n_estimators": [100, 180],
                    "learning_rate": [0.03, 0.05, 0.1],
                    "num_leaves": [31, 63],
                    "subsample": [0.8, 1.0],
                    "colsample_bytree": [0.8, 1.0],
                },
                n_iter=3,
                cv=tscv,
                scoring="neg_root_mean_squared_error",
                random_state=SEED,
                n_jobs=1,
            )
            lgbm_search.fit(X_train, y_train)
            t1 = time.perf_counter()
            lgbm_model = lgbm_search.best_estimator_
            pred_lgbm = lgbm_model.predict(X_test)
            predictions["Boosting_LightGBM"] = pred_lgbm
            metrics_rows["Boosting_LightGBM"] = compute_metrics(y_test, pred_lgbm, t1 - t0)
            trained_models["lightgbm"] = lgbm_model
            tqdm.write(f"[5/8] LightGBM done in {t1 - t0:.2f}s | Best: {lgbm_search.best_params_}")

        # 6. Boosting C: CatBoost
        elif step_name.startswith("6/8"):
            t0 = time.perf_counter()
            cb_model = CatBoostRegressor(
                iterations=250,
                learning_rate=0.06,
                depth=6,
                random_seed=SEED,
                verbose=0,
            )
            cb_model.fit(X_train, y_train)
            t1 = time.perf_counter()
            pred_cb = cb_model.predict(X_test)
            predictions["Boosting_CatBoost"] = pred_cb
            metrics_rows["Boosting_CatBoost"] = compute_metrics(y_test, pred_cb, t1 - t0)
            trained_models["catboost"] = cb_model
            tqdm.write(f"[6/8] CatBoost done in {t1 - t0:.2f}s")

        # 7. Stacking: RF + XGBoost + LightGBM + CatBoost -> Ridge meta-model (time-series safe)
        elif step_name.startswith("7/8"):
            t0 = time.perf_counter()
            base_estimators = {
                "rf": RandomForestRegressor(n_estimators=60, max_depth=12, random_state=SEED, n_jobs=2),
                "xgb": XGBRegressor(n_estimators=120, learning_rate=0.05, max_depth=6, random_state=SEED, n_jobs=2, tree_method="hist"),
                "lgbm": LGBMRegressor(n_estimators=120, learning_rate=0.05, num_leaves=31, random_state=SEED, n_jobs=2, verbose=-1),
                "cb": CatBoostRegressor(iterations=200, learning_rate=0.06, depth=6, random_seed=SEED, verbose=0),
            }
            stack_bundle, pred_stack = fit_timeseries_stacker(base_estimators, X_train, y_train, X_test, tscv)
            t1 = time.perf_counter()
            predictions["Stacking_Ensemble"] = pred_stack
            metrics_rows["Stacking_Ensemble"] = compute_metrics(y_test, pred_stack, t1 - t0)
            trained_models["stacking"] = stack_bundle
            tqdm.write(f"[7/8] Stacking Ensemble done in {t1 - t0:.2f}s")

        # 8. Hybrid Model: ARIMA + Random Forest Residual Forecast
        elif step_name.startswith("8/8"):
            t0 = time.perf_counter()
            arima_order = (2, 1, 2)
            arima_series = pd.Series(y_train[-min(len(y_train), 2500):])
            arima_fit = ARIMA(arima_series, order=arima_order).fit(method_kwargs={"warn_convergence": False})

            in_sample_fitted = arima_fit.fittedvalues.to_numpy(dtype=float)
            actual_win = arima_series.to_numpy(dtype=float)
            train_residuals = actual_win - in_sample_fitted

            X_res_train = X_train.iloc[-len(arima_series):].copy()
            rf_residual = RandomForestRegressor(
                n_estimators=80,
                max_depth=12,
                random_state=SEED,
                n_jobs=2,
            )
            rf_residual.fit(X_res_train, train_residuals)

            arima_test_res = arima_fit.apply(y_test)
            arima_test_pred = np.asarray(arima_test_res.fittedvalues, dtype=float)
            if len(arima_test_pred) != len(y_test):
                arima_test_pred = np.asarray(arima_fit.forecast(steps=len(y_test)), dtype=float)

            pred_residual = rf_residual.predict(X_test)
            pred_hybrid = arima_test_pred + pred_residual
            t1 = time.perf_counter()

            predictions["Hybrid_ARIMA_RF"] = pred_hybrid
            metrics_rows["Hybrid_ARIMA_RF"] = compute_metrics(y_test, pred_hybrid, t1 - t0)
            trained_models["hybrid_arima_fit"] = arima_fit
            trained_models["hybrid_rf_residual"] = rf_residual
            tqdm.write(f"[8/8] Hybrid ARIMA+RF done in {t1 - t0:.2f}s")

    metrics_df = pd.DataFrame.from_dict(metrics_rows, orient="index")
    metrics_df.index.name = "Model"
    metrics_df.to_csv(artifact_dir / "model_metrics.csv")

    preds_df = pd.DataFrame(predictions, index=test_df.index)
    preds_df["Actual"] = y_test
    preds_df.to_csv(artifact_dir / "test_predictions.csv")

    joblib.dump(
        {
            "models": trained_models,
            "feature_cols": feature_cols,
            "target_col": target_col,
            "arima_order": arima_order,
        },
        artifact_dir / "ensemble_bundle.joblib",
    )

    print("\n--- Step 4 Test Metrics Preview ---")
    print(metrics_df.to_string())
    return metrics_df, preds_df


if __name__ == "__main__":
    run_models()
```

```python
# pip install pandas numpy matplotlib scikit-learn xgboost joblib shap

from pathlib import Path
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.inspection import permutation_importance

CLEAN_PATH = Path("data/processed/sl_electricity_load_clean.csv")
ARTIFACT_DIR = Path("artifacts/models")
FIG_DIR = Path("reports/figures")
SEED = 42


def run_explainability(
    clean_path: Path = CLEAN_PATH,
    artifact_dir: Path = ARTIFACT_DIR,
    fig_dir: Path = FIG_DIR,
) -> pd.DataFrame:
    bundle_path = artifact_dir / "ensemble_bundle.joblib"
    if not bundle_path.exists():
        raise FileNotFoundError(f"Bundle not found at '{bundle_path}'. Run Step 4 first.")
    if not clean_path.exists():
        raise FileNotFoundError(f"Cleaned CSV not found at '{clean_path}'. Run Step 2 first.")

    fig_dir.mkdir(parents=True, exist_ok=True)
    bundle = joblib.load(bundle_path)
    models = bundle["models"]
    feature_cols = bundle["feature_cols"]
    target_col = bundle["target_col"]

    df = pd.read_csv(clean_path, index_col=0, parse_dates=True).sort_index()
    split_idx = int(len(df) * 0.8)
    X_test = df.iloc[split_idx:][feature_cols].copy()
    y_test = df.iloc[split_idx:][target_col].to_numpy(dtype=float)

    importance_df = pd.DataFrame({"feature": feature_cols})

    # 1. Built-in Random Forest importance
    if "random_forest" in models and hasattr(models["random_forest"], "feature_importances_"):
        rf_imp = np.asarray(models["random_forest"].feature_importances_, dtype=float)
        importance_df["rf_importance"] = rf_imp / (rf_imp.sum() + 1e-12)

        top_rf = importance_df.sort_values("rf_importance", ascending=False).head(12)
        fig, ax = plt.subplots(figsize=(9, 4.8))
        ax.barh(top_rf["feature"][::-1], top_rf["rf_importance"][::-1], color="#1f77b4")
        ax.set_title("Top 12 Features — Random Forest (Normalized Importance)")
        ax.set_xlabel("Relative Importance")
        fig.tight_layout()
        fig.savefig(fig_dir / "09_rf_feature_importance.png", dpi=150)
        plt.close(fig)

    # 2. Built-in XGBoost importance
    if "xgboost" in models and hasattr(models["xgboost"], "feature_importances_"):
        xgb_imp = np.asarray(models["xgboost"].feature_importances_, dtype=float)
        importance_df["xgb_importance"] = xgb_imp / (xgb_imp.sum() + 1e-12)

        top_xgb = importance_df.sort_values("xgb_importance", ascending=False).head(12)
        fig, ax = plt.subplots(figsize=(9, 4.8))
        ax.barh(top_xgb["feature"][::-1], top_xgb["xgb_importance"][::-1], color="#ff7f0e")
        ax.set_title("Top 12 Features — XGBoost (Normalized Importance)")
        ax.set_xlabel("Relative Importance")
        fig.tight_layout()
        fig.savefig(fig_dir / "10_xgb_feature_importance.png", dpi=150)
        plt.close(fig)

    # 3. SHAP values (on up to 1,000 test rows for speed), with permutation fallback
    sample_n = min(1000, len(X_test))
    rng = np.random.default_rng(SEED)
    sample_idx = np.sort(rng.choice(len(X_test), size=sample_n, replace=False))
    X_sample = X_test.iloc[sample_idx].copy()
    y_sample = y_test[sample_idx]

    explainer_model = models.get("xgboost") or models.get("random_forest")
    shap_computed = False

    if explainer_model is not None:
        try:
            import shap

            explainer = shap.TreeExplainer(explainer_model)
            shap_values = explainer.shap_values(X_sample)
            shap_arr = np.asarray(shap_values, dtype=float)
            mean_abs_shap = np.mean(np.abs(shap_arr), axis=0)
            importance_df["mean_abs_shap"] = mean_abs_shap
            shap_computed = True

            # SHAP summary bar plot
            fig, ax = plt.subplots(figsize=(9, 4.8))
            shap_order = np.argsort(mean_abs_shap)[-12:]
            ax.barh(
                np.array(feature_cols)[shap_order],
                mean_abs_shap[shap_order],
                color="#2ca02c",
            )
            ax.set_title("Top 12 Features — Mean |SHAP Value| (Impact on Forecast)")
            ax.set_xlabel("Mean |SHAP Value|")
            fig.tight_layout()
            fig.savefig(fig_dir / "11_shap_summary_bar.png", dpi=150)
            plt.close(fig)

            # SHAP beeswarm / dot plot
            plt.figure(figsize=(9, 5))
            shap.summary_plot(shap_arr, X_sample, max_display=12, show=False)
            plt.tight_layout()
            plt.savefig(fig_dir / "12_shap_beeswarm.png", dpi=150, bbox_inches="tight")
            plt.close()
        except Exception as exc:
            print(f"SHAP skipped ({exc}); using permutation importance fallback.")

    if not shap_computed and explainer_model is not None:
        perm = permutation_importance(
            explainer_model,
            X_sample,
            y_sample,
            n_repeats=5,
            random_state=SEED,
            n_jobs=2,
        )
        importance_df["permutation_importance"] = np.maximum(perm.importances_mean, 0.0)

    sort_col = (
        "mean_abs_shap"
        if "mean_abs_shap" in importance_df.columns
        else "rf_importance"
        if "rf_importance" in importance_df.columns
        else "xgb_importance"
    )
    importance_df = importance_df.sort_values(sort_col, ascending=False).reset_index(drop=True)
    importance_df.to_csv(artifact_dir / "feature_importance.csv", index=False)

    print("\n--- Top 10 Most Important Features ---")
    print(importance_df.head(10).to_string(index=False))
    return importance_df


if __name__ == "__main__":
    run_explainability()
```
